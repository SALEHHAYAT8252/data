# Question 7: Multiple Linear Regression
To implement Multiple Linear Regression using Python and predict students' exam scores based on study hours, attendance and assignment scores.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 1: Create the dataset
data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9, 4, 6],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95, 68, 82],
    "Assignment_Score": [55, 60, 65, 70, 75, 80, 85, 90, 62, 78],
    "Exam_Score": [45, 50, 58, 65, 72, 78, 86, 92, 55, 75]
}

df = pd.DataFrame(data)

print("Student Dataset:")
print(df)
```

```python
# Step 2: Separate input and output variables
X = df[["Study_Hours", "Attendance", "Assignment_Score"]]
y = df["Exam_Score"]
```

```python
# Step 3: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# Step 4: Create and train the model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Step 5: Predict exam scores
y_pred = model.predict(X_test)

print("\nActual Exam Scores:")
print(y_test.tolist())

print("\nPredicted Exam Scores:")
print(y_pred.round(2))
```

```python
# Step 6: Evaluate the model
print("\nMean Squared Error:",
      round(mean_squared_error(y_test, y_pred), 2))

print("R2 Score:", round(r2_score(y_test, y_pred), 2))
```

```python
# Step 7: Display coefficients and intercept
print("\nCoefficients:", model.coef_)
print("Intercept:", round(model.intercept_, 2))
```

```python
# Step 8: Predict for a new student
new_student = pd.DataFrame({
    "Study_Hours": [7],
    "Attendance": [85],
    "Assignment_Score": [82]
})

predicted_score = model.predict(new_student)

print("\nPredicted Exam Score for New Student:",
      round(predicted_score[0], 2))
```
