### Question: 2
Write a Pandas program to perform the following operations on a CSV file (students.csv) stored in local system:

1. Read the CSV file and display first 5 rows.
2. Display last 3 rows and dataset information.
3. Select and display specific columns (Name, Marks).
4. Filter students who scored more than 75 marks.
5. Add a new column Grade based on marks.
6. Check and handle missing values.
7. Sort data based on marks in descending order.
8. Perform group by operation to find average marks by Department.
9. Save the updated dataset into a new CSV file.
10. Display statistical summary of the dataset.


```python
# 1. Read CSV file and display first 5 rows
df = pd.read_csv("students.csv")
print("First 5 rows:\n", df.head())
```

```python
# 2. Display last 3 rows and dataset info
print("\nLast 3 rows:\n", df.tail(3))
print("\nDataset Info:")
print(df.info())
```

```python
# 3. Select specific columns
print("\nSelected Columns (Name, Marks):\n", df[['Name', 'Marks']])
```

```python
# 4. Filter students with marks > 75
high_marks = df[df['Marks'] > 75]
print("\nStudents with Marks > 75:\n", high_marks)
```

```python
# 5. Add Grade column
def grade(marks):
    if marks >= 80:
        return 'A'
    elif marks >= 60:
        return 'B'
    else:
        return 'C'

df['Grade'] = df['Marks'].apply(grade)
print("\nData with Grade:\n", df)
```

```python
# 6. Handle missing values
print("\nMissing Values:\n", df.isnull().sum())
df.fillna(0, inplace=True)
```

```python
# 7. Sort data by Marks (descending)
sorted_df = df.sort_values(by='Marks', ascending=False)
print("\nSorted Data:\n", sorted_df)
```

```python
# 8. Group by Department and find average marks
grouped = df.groupby('Department')['Marks'].mean()
print("\nAverage Marks by Department:\n", grouped)
```

```python
# 9. Save updated data to new CSV
df.to_csv("Students.csv", index=False)
```

```python
# 10. Statistical summary
print("\nStatistical Summary:\n", df.describe())
```

### Question: 3
Write a Python program using Pandas to analyze the students.csv dataset and perform the following operations:

1. Read the students.csv file.
2. Display the total number of students.
3. Find the student who obtained the highest marks.
4. Find the student who obtained the lowest marks.
5. Calculate the average, maximum, and minimum marks.
6. Count the number of students in each Department.
7. Count the number of students in each Grade.
8. Display students whose marks are between 60 and 80.
9. Display the names of students belonging to the CSE department.
10. Create a new CSV file containing only students who scored 75 or more marks.

```python
# 1. Read the CSV file
df = pd.read_csv("Students.csv")

print("Student Dataset:\n")
print(df)
```

```python
# 2. Display total number of students
print("\nTotal Number of Students:", len(df))
```

```python
# 3. Find student with highest marks
highest = df.loc[df['Marks'].idxmax()]
print("\nStudent with Highest Marks:")
print(highest)
```

```python
# 4. Find student with lowest marks
lowest = df.loc[df['Marks'].idxmin()]
print("\nStudent with Lowest Marks:")
print(lowest)
```

```python
# 5. Calculate average, maximum and minimum marks
print("\nAverage Marks:", df['Marks'].mean())
print("Maximum Marks:", df['Marks'].max())
print("Minimum Marks:", df['Marks'].min())
```

```python
# 6. Count students in each Department
print("\nNumber of Students in Each Department:")
print(df['Department'].value_counts())
```

```python
# 7. Count students in each Grade
print("\nNumber of Students in Each Grade:")
print(df['Grade'].value_counts())
```

```python
# 8. Display students with marks between 60 and 80
students_60_80 = df[df['Marks'].between(60, 80)]

print("\nStudents with Marks Between 60 and 80:")
print(students_60_80)
```

```python
# 9. Display students belonging to CSE department
cse_students = df[df['Department'] == 'CSE']

print("\nCSE Students:")
print(cse_students[['Name', 'Marks', 'Age', 'Grade']])
```

```python
# 10. Save students scoring 75 or more
top_students = df[df['Marks'] >= 75]

top_students.to_csv("students_75_plus.csv", index=False)

print("\nstudents_75_plus.csv file created successfully!")
```

### Question: 4
Write a Python program using the Pandas library to work with two CSV files, students.csv and fees.csv. Perform the following operations:

1. Read both CSV files into separate DataFrames.
2. Display the contents of both DataFrames.
3. Merge the two DataFrames using the common Name column.
4. Display only the Name, Department, Marks, and Fees_Status columns.
5. Find students whose fees are Pending.
6. Find students who scored more than 75 marks and have Paid fees.
7. Count the number of students according to Fees_Status.
8. Calculate the average marks of students for each Fees_Status.
9. Sort the merged DataFrame according to Marks in descending order.
10. Save the merged DataFrame as student_fee_details.csv.

```python
import pandas as pd

fees_data = {
    "Name": ["Amit", "Riya", "Rahul", "Sneha", "Arjun",
             "Pooja", "Karan", "Neha", "Vikas", "Anjali"],
    "Fees_Status": ["Paid", "Pending", "Paid", "Paid", "Pending",
                    "Paid", "Paid", "Pending", "Pending", "Paid"]
}

fees_df = pd.DataFrame(fees_data)

fees_df.to_csv("fees.csv", index=False)

print("fees.csv file created successfully!")
```

```python
# 1) Read both CSV files
students = pd.read_csv("students.csv")
fees = pd.read_csv("fees.csv")
```

```python
# 2) Display both DataFrames
print("Students Data:")
print(students)

print("\nFees Data:")
print(fees)
```

```python
# 3) Merge both DataFrames using Name
merged_df = pd.merge(students, fees, on="Name")

print("\nMerged Data:")
print(merged_df)
```

```python
# 4) Display selected columns
print("\nSelected Columns:")
print(merged_df[["Name", "Department", "Marks", "Fees_Status"]])
```

```python
# 5) Find students whose fees are Pending
pending = merged_df[merged_df["Fees_Status"] == "Pending"]

print("\nStudents with Pending Fees:")
print(pending[["Name", "Department", "Marks", "Fees_Status"]])
```

```python
# 6) Students with marks > 75 and Paid fees
top_paid = merged_df[
    (merged_df["Marks"] > 75) &
    (merged_df["Fees_Status"] == "Paid")
]

print("\nStudents with Marks > 75 and Paid Fees:")
print(top_paid[["Name", "Marks", "Fees_Status"]])
```

```python
# 7) Count students according to Fees_Status
print("\nFees Status Count:")
print(merged_df["Fees_Status"].value_counts())
```

```python
# 8) Average marks according to Fees_Status
print("\nAverage Marks by Fees Status:")
print(merged_df.groupby("Fees_Status")["Marks"].mean())
```

```python
# 9) Sort by Marks in descending order
sorted_df = merged_df.sort_values(
    by="Marks",
    ascending=False
)

print("\nData Sorted by Marks:")
print(sorted_df)
```

```python
# 10) Save merged DataFrame
merged_df.to_csv(
    "student_fee_details.csv",
    index=False
)

print("\nstudent_fee_details.csv file created successfully!")
```
