## Question: 5
Create a dataset of 30 students with the following attributes:

Student_ID
Age
Study_Hours
Attendance
Assignment_Score
Exam_Score

Using Python, Pandas, NumPy, and SciPy, perform the following statistical operations on the dataset:
1. Calculate Mean, Median, and Mode for the numerical variables.
2. Calculate Minimum, Maximum, Range, and Sum.
3. Calculate Variance and Standard Deviation.
4. Calculate Quartiles (Q1, Q2, Q3) and Interquartile Range (IQR).
5. Calculate 25th, 50th, 75th, and 90th Percentiles.
6. Calculate Skewness and Kurtosis.
7. Calculate the Covariance between Study_Hours and Exam_Score.
8. Calculate the Pearson Correlation between Study_Hours and Exam_Score.
9. Identify outliers using the IQR method.
10. Display the complete statistical summary using describe() and visualize the data using a Histogram, Box Plot, and Scatter Plot.

```python
import pandas as pd 
import numpy as np
from scipy import stats 
import matplotlib.pyplot as plt 

# Create dataset of 30 students 

np.random.seed(42)
df = pd.DataFrame({
    'Student_ID': range(1, 31), 
    'Age': np.random.randint(18, 24, 30),
    'Study_Hours': np.random.randint(2, 10, 30),
    'Attendance': np.random.randint(60, 101, 30),
    'Assignment_Score': np.random.randint(50, 101, 30),
    'Exam_Score': np.random.randint(45, 101, 30) })
print(df)
```

```python
# 1. Mean, Median and Mode
print("Mean:\n", df.mean(numeric_only=True))
print("Median:\n", df.median(numeric_only=True))
print("Mode:\n", df.mode(numeric_only=True).iloc[0])
```

```python
# 2. Minimum, Maximum, Range and Sum
print("Minimum:\n", df.min(numeric_only=True))
print("Maximum:\n", df.max(numeric_only=True))
print("Range:\n", df.max(numeric_only=True) - df.min(numeric_only=True))
print("Sum:\n", df.sum(numeric_only=True))
```

```python
# 3. Variance and Standard Deviation
print("Variance:\n", df.var(numeric_only=True))
print("Standard Deviation:\n", df.std(numeric_only=True))
```

```python
# 4. Quartiles and IQR
Q1 = df['Exam_Score'].quantile(0.25)
Q2 = df['Exam_Score'].quantile(0.50)
Q3 = df['Exam_Score'].quantile(0.75)
IQR =Q3 - Q1 
print("Q1, Q2, Q3:", Q1, Q2, Q3)
print("IQR:", IQR)
```

```python
# 5. Percentiles
print("Percentiles:\n", np.percentile(df['Exam_Score'], [25, 50, 75, 90]))
```

```python
# 6. Skewness and Kurtosis
print("Skewness:\n", df.skew(numeric_only=True))
print("Kurtosis:\n", df.kurt(numeric_only=True))
```

```python
# 7. Covariance
print("Covariance:\n", df[['Study_Hours', 'Exam_Score']].cov())
```

```python
# 8. Pearson Correlation
print("Pearson Correlation:", df['Study_Hours'].corr(df['Exam_Score']))
```

```python
# 9. Outliers using IQR
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['Exam_Score'] < lower) | (df['Exam_Score'] > upper)]
print("Outliers:\n", outliers)
```

```python
# 10. Statistical summary and visualizations
print("Statistical Summary:\n", df.describe())

plt.hist(df['Exam_Score'], bins=6)
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.title('Histogram of Exam Scores')
plt.show()

plt.boxplot(df['Exam_Score'])
plt.ylabel('Exam Score')
plt.title('Box Plot of Exam Scores')
plt.show()

plt.scatter(df['Study_Hours'], df['Exam_Score'])
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.show()
```
